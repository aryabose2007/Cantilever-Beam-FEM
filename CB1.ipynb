```python
import numpy as np
from sympy import *
```

```python
import matplotlib.pyplot as plt
```

```python
L=5
b= 0.25
h = 0.85 ##all in metres
E = 2*10**11 ##E of structural steel in ansys is 200 Gpa
P = 3000 #in Newton
```

```python
I = (b*h**3)/12  ##second moment of area
```

```python
x = np.linspace(0,L,100)
M = P*(L-x) ##bending moment 
```

```python
v= P*x**2*(3*L-x)/6*E*I ##deflection along beam
```

```python
𝛉 = P*x*(2*L-x)/2*E*I  ##slope of beam
```

```python
𝛔 = (P*(L-x)*h)/2*I
```

```python
𝛜 = 𝛔/E
```

```python
V = -P*np.ones_like(x) ##sheer internal force varying conditon
```

```python
##PLotting deflection along beam
plt.figure()
plt.plot(x,v)
plt.title('Plot of Deflection of Beam')
plt.xlabel('x=Distance from fixed end to the point on beam')
plt.ylabel('v= deflection along the beam')
plt.grid(color = 'm', linewidth = 0.4)
plt.show
```

```python
##PLotting Bending Moment along beam
plt.figure()
plt.plot(x,M)
plt.title('Plot of Bending Moment along Beam')
plt.xlabel('x=Distance from fixed end to the point on beam')
plt.ylabel('M= Bending Moment along the beam')
plt.grid(color = 'b', linewidth = 0.4)
plt.show
```

```python
##PLotting sheer internal force along beam
plt.figure()
plt.plot(x,V)
plt.title('Plot of Sheer internal force along Beam')
plt.xlabel('x=Distance from fixed end to the point on beam')
plt.ylabel('V= sheer internal force along the beam')
plt.grid(color = 'black', linewidth = 0.4)
plt.show
```

```python
##PLotting sheer stress along beam
plt.figure()
plt.plot(x,𝛔 ,color = 'black', linewidth = 0.5 )
plt.title('Plot of sheer stress of Beam')
plt.xlabel('x=Distance from fixed end to the point on beam')
plt.ylabel('𝛔 = sheer stress along the beam')
plt.grid(color = 'green', linewidth = 0.4)
plt.show
```

```python
##PLotting slope along beam
plt.figure()
plt.plot(x,𝛉, color = 'magenta', linewidth = 0.4)
plt.title('Plot of slope of deflection along Beam')
plt.xlabel('x=Distance from fixed end to the point on beam')
plt.ylabel('𝛉= slope(rad) along the beam')
plt.grid(color = 'brown', linewidth = 0.5)
plt.show
```

```python
##PLotting value of young's modulus
plt.figure()
plt.plot(𝛜,𝛔, color = 'magenta', linewidth = 0.4)
plt.title('Plotting the value of Young modulus')
plt.xlabel('Strain')
plt.ylabel('Stress')
plt.grid(color = 'brown', linewidth = 0.5)
plt.show
```
